QS-5

```python
correct_pin = "1234"
attempts = 0
while attempts < 4:
    pin = input("Enter your PIN: ")
    if pin == correct_pin:
        print("Access Granted")
        break
    attempts = attempts + 1
if attempts == 4:
    print("Card Blocked")
```

QS-4

```python
n = int(input("Enter number of students: "))
present = 0
absent = 0
for i in range(n):
    attendance = int(input("Enter attendance (1 = Present, 0 = Absent): "))
    if attendance == 1:
        present = present + 1
    elif attendance == 0:
        absent = absent + 1
print("Number of students present:", present)
print("Number of students absent:", absent)
```
