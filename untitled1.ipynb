```python
n = int(input("Enter number of students: "))
marks_list = []
for i in range(1, n + 1):
    marks = float(input(f"Enter marks of student {i}: "))
    marks_list.append(marks)
highest = marks_list[0]
lowest = marks_list[0]
for marks in marks_list:
    if marks > highest:
        highest = marks
    if marks < lowest:
        lowest = marks
print("Highest marks =", highest)
print("Lowest marks =", lowest)
```

```python
account_number = int(input("Enter account number: "))
reversed_number = 0
while account_number > 0:
    digit = account_number % 10
    reversed_number = reversed_number * 10 + digit
    account_number //= 10
print("Reversed account number =", reversed_number)
```

```python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi
for person in range(1, 6):
    weight = float(input(f"Enter weight (kg) for person {person}: "))
    height = float(input(f"Enter height (m) for person {person}: "))
    bmi = calculate_bmi(weight, height)
    print(f"BMI of person {person} =", bmi)
```

QS-8

```python
def is_prime(number):
    if number < 2:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True
print("Prime numbers between 1 and 100:")
for num in range(1, 101):
    if is_prime(num):
        print(num)
```

QS-9

```python
def calculate_salary(hours, rate):
    if hours <= 40:
        salary = hours * rate
    else:
        overtime_hours = hours - 40
        salary = (40 * rate) + (overtime_hours * rate * 1.5)
    return salary
hours = float(input("Enter hours worked (0 to stop): "))
while hours != 0:
    rate = float(input("Enter hourly rate: "))
    salary = calculate_salary(hours, rate)
    print("Salary =", salary)
    hours = float(input("Enter hours worked (0 to stop): "))
```
